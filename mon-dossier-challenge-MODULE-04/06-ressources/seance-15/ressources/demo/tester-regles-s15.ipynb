## 1. Comparer A et C corrigée
**À prédire.** L’une de ces versions dépasse-t-elle le seuil de l’exercice ? Quel passage du guide S12 vous permet de vérifier le fait conservé ?

Dans les deux formulations de S14, le dossier conserve V1 et V2 et le journal explique les corrections. Relevez dans la source le passage de l’étape 5 du guide et dans le journal la ligne « Forme — accessibilité ». La demande d’extrait reste identique. Les mots employés pour présenter le journal changent.

La fonction `len` compte les points de code Unicode du texte Python. Espaces, ponctuation et retours à la ligne comptent. La présence est contrôlée séparément.

```python
cas = {'source': 'Produire avec l’IA sous responsabilité humaine — Guide pratique, V2 à '
           'relire. Étape 5 du guide et journal de corrections S12.',
 'public': 'Formateurs débutants qui relisent un premier support produit avec l’IA.',
 'fait1': 'Le dossier conserve les versions V1 et V2 du guide.',
 'fait2': 'Le journal explique les corrections et leurs raisons.',
 'variable': 'La formulation de la proposition',
 'action': 'Souhaitez-vous voir un extrait du journal de corrections ?',
 'messageA': 'Le dossier conserve V1 et V2 du guide. Le journal explique chaque '
             'correction et sa raison. Souhaitez-vous voir un extrait du journal de '
             'corrections ?',
 'messageC': 'Le dossier conserve les versions V1 et V2 du guide. Le journal explique ce '
             'qui a changé et pourquoi. Souhaitez-vous voir un extrait du journal de '
             'corrections ?',
 'seuil': 400,
 'ok': 'Longueur acceptable',
 'alerte': 'À raccourcir'}

def verdict_longueur(texte, cas):
    """Appliquer la condition du seuil à un texte fourni."""
    if len(texte) > cas["seuil"]:
        return cas["alerte"]
    return cas["ok"]


def ligne_mesure(nom, texte, cas):
    """Vérifier la présence avant de rendre un avis sur la longueur."""
    if not texte.strip():
        return f"{nom} | {len(texte)} caractère(s) | À compléter | longueur non évaluée"
    return f"{nom} | {len(texte)} caractères | {verdict_longueur(texte, cas)}"


print("1. Retrouver les deux variantes de S14")
print(cas["source"])
print("Public :", cas["public"])
print("Faits à conserver :", cas["fait1"], cas["fait2"])
print("Élément variable :", cas["variable"])
for nom, texte in [("A", cas["messageA"]), ("C corrigée", cas["messageC"])]:
    print(nom, "·", texte)
    print(ligne_mesure(nom, texte, cas))
print("Action commune :", cas["action"])
```

## 2. Construire un gabarit

**À prédire.** Si vous changez le champ `proposition`, quelle partie du message doit rester identique ?

Les accolades désignent trois emplacements. La méthode `format_map` remplit le gabarit à partir des champs nommés. Changez la proposition avec une formulation appuyée sur le journal, puis relancez la cellule. Conservez la même demande au lecteur.

```python
cas.update({'gabarit': '{titre} — {proposition} {action}',
 'gabaritChamps': {'titre': 'Révision du guide',
                   'proposition': 'Le journal explique ce qui a changé et pourquoi.',
                   'action': 'Souhaitez-vous voir un extrait du journal de corrections '
                             '?'}})

print("\n2. Assembler un gabarit")
modele = cas["gabarit"]
champs = cas["gabaritChamps"]
print("Gabarit :", modele)
for nom in ("titre", "proposition", "action"):
    print(nom, "=", champs[nom])
message = modele.format_map(champs)
print(message)
print(ligne_mesure("Gabarit complété", message, cas))
```

## 3. Prédire à 400 puis à 401 caractères

**Avant de lancer la cellule**, écrivez les deux verdicts prévus. Expliquez la différence entre `>` et `>=`.

Le motif « 0123456789 » contient dix caractères. Sa répétition quarante fois fournit un texte étalon de 400 caractères. Ajoutez le caractère X pour atteindre 401. Ce sont des données d’essai. Le seuil de 400 appartient à cette séance.

Pour un texte vide, le contrôle de présence demande de compléter le champ. Le verdict de longueur reste en attente.

```python
cas.update({'etalon': '0123456789', 'suffixe': 'X'})

print("\n3. Prédire puis tester la frontière")
# Le motif lisible compte dix caractères. Répété quarante fois, il en compte 400.
texte_400 = cas["etalon"] * 40
texte_401 = texte_400 + cas["suffixe"]
if len(texte_400) != 400 or len(texte_401) != 401:
    raise ValueError("Pour cet exercice, gardez un motif de 10 caractères et un suffixe de 1.")
print("Seuil d’exercice :", cas["seuil"], "caractères.")
print("Si la longueur dépasse le seuil, afficher «", cas["alerte"], "».")
print("Sinon, afficher «", cas["ok"], "».")
print(ligne_mesure("Test 1 · motif × 40", texte_400, cas))
print(ligne_mesure("Test 2 · même texte + " + cas["suffixe"], texte_401, cas))
print(ligne_mesure("Essai de présence", "", cas))
assert len(texte_400) == 400 and len(texte_401) == 401
if cas["seuil"] == 400:
    assert verdict_longueur(texte_400, cas) == cas["ok"]
    assert verdict_longueur(texte_401, cas) == cas["alerte"]
assert "À compléter" in ligne_mesure("Vide", "", cas)
```
